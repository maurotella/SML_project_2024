```python
import os
import pandas as pd
import yaml as YAML
```

```python
run_names = os.listdir('dataset')
all = pd.DataFrame(data={'path':[],'ate':[],'are':[],'area':[]})

for run in run_names:
    path = f'dataset/{run}'
    n = max([int(x[:x.find('APE')]) for x in os.listdir(path) if x.endswith('.yaml')])
    for i in range(n+1):
        with open(f'{path}/{i}APE.yaml') as yaml:
            APE = YAML.safe_load(yaml)
            #print(APE)
            #print(f'{path}/{i}APE.yaml')
            del APE['ape']
            new = pd.DataFrame(
                dict({'path': [f'{path}/{i}{".0" if i else ""}MapBounded.png']},**APE)
            )
        all = pd.concat([all, new],ignore_index=True)
all.to_csv('annotation_file.csv',index=False)
```
